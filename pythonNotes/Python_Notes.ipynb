### 1.1 Strings


```python
# A string is defined with quotes. Indexes start at 0. Negative indexes count from the end.
name = "The BodyGuard"
print(name[0])          # first character
print(name[-1])         # last character
print(name[0:4])        # slice includes the start and excludes the end
print(name[::2])        # stride of 2: take every other character
print(name + " is the best album")
print(3 * "ab")

# Escapes: \n is a newline and \t is a tab. An r prefix makes a raw string, so backslashes are literal.
print("line1\nline2")
print("col1\tcol2")
print(r"C:\new\folder")

title1 = "data science basics"
print(title1.title())   # capitalize each word

a = "Thriller is the sixth studio album"
print("before upper:", a)
b = a.upper()
print("after upper:", b)
print(a.replace("Thriller", "Janet"))
print(a.find("sixth"))          # index where the substring starts
print(a.find("missing"))        # -1 when the substring is absent
print(a.split())                # split on whitespace into a list
```

```python
import re

# search finds the first match, findall finds all of them, sub replaces, and split cuts on a pattern.
s1 = "The BodyGuard is the best album"
print(re.search(r"Body", s1).group())

text = "My Phone number is 1234567890"
match = re.search(r"\d{10}", text)   # \d matches a digit; {10} means ten in a row
if match:
    print("Phone number found:", match.group())

print("non-word chars:", re.findall(r"\W", "Hello, world!"))
print(re.split(r"\s", s1))
print(re.sub(r"fox", "bear", "The quick brown fox jumps over the lazy dog."))
```

### 1.2 Lists, tuples, sets, and dictionaries


```python
# Lists can be changed. Indexing and slicing work the same way as with strings.
A = ["test", 1, 2]
A.append(3)
print(A)
print(A[0], A[-1], len(A))
print(type(A))
# help(A) prints every list method. It is left out here so the notes stay short.

# A tuple cannot replace its items, but a list stored inside it can still be changed.
tuple1 = (2, [2, 30], "txt")
print(tuple1, type(tuple1))
tuple1[1].append(40)
print(tuple1)

# A set drops duplicates and is useful for intersection, union, and difference.
album_set = {"Thriller", "Back in Black", "AC/DC", "Thriller"}
album_set.add("NSYNC")
print(album_set)
print(album_set & {"Thriller", "Dark Side"})   # intersection
print(album_set | {"Dark Side"})               # union

# A dictionary looks up values by key. Keys are usually immutable.
release = {"Thriller": 1982, "Back in Black": 1980}
release["The BodyGuard"] = 1992
print(release["Thriller"])
print(list(release.keys()))
print("Thriller" in release)
del release["Back in Black"]
print(release)
```

```python
age = 18
if age > 18:
    print("adult")
elif age == 18:
    print("just 18")
else:
    print("minor")

# range includes the start and excludes the end. This prints 1 through 4.
for i in range(1, 5):
    print(i)

squares = []
i = 1
while i < 5:
    squares.append(i ** 2)
    i += 1
print(squares)

# A list comprehension is a short form of the loop above.
print([i ** 2 for i in range(1, 5) if i % 2 == 1])
```

## 2. Data Analysis with Python

This section follows the lab order: load a table with missing values, clean it, explore it, fit a linear regression, then split the data and tune the model.

The table is a small stand-in for the car-price lab, so it does not need the course CSV. The columns are related on purpose, which lets the regression show how `engine-size`, `highway-mpg`, and `horsepower` connect to `price`.


### 2.1 Build a table that contains missing values


```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(0)
n = 120
engine = rng.normal(150, 40, n)
mpg = rng.normal(30, 6, n)
hp = rng.normal(110, 30, n)
price = 180 * engine - 350 * mpg + 60 * hp + rng.normal(0, 1200, n)

df_raw = pd.DataFrame({
    "engine-size": engine.round(0),
    "highway-mpg": mpg.round(1),
    "horsepower": hp.round(1),
    "peak-rpm": rng.normal(5100, 400, n).round(0),
    "bore": rng.normal(3.3, 0.25, n).round(2),
    "stroke": rng.normal(3.3, 0.2, n).round(2),
    "num-of-doors": rng.choice(["two", "four"], n),
    "fuel-type": rng.choice(["gas", "diesel"], n, p=[0.85, 0.15]),
    "aspiration": rng.choice(["std", "turbo"], n, p=[0.8, 0.2]),
    "body-style": rng.choice(["sedan", "hatchback", "wagon", "hardtop"], n),
    "drive-wheels": rng.choice(["fwd", "rwd", "4wd"], n),
    "engine-location": rng.choice(["front", "rear"], n, p=[0.92, 0.08]),
    "price": price.round(0),
})

# Insert "?" and NaN, matching the lab step that calls replace("?", np.nan) and then fills the gaps.
# Convert to object first. A float column cannot store a question mark.
df_raw["horsepower"] = df_raw["horsepower"].astype(object)
df_raw.loc[0:4, "horsepower"] = "?"
df_raw.loc[5:8, "stroke"] = np.nan
df_raw.loc[9:11, "num-of-doors"] = np.nan
df_raw.loc[12:16, "price"] = np.nan
df_raw.loc[17:18, "bore"] = np.nan
df_raw.head(8)
```

### 2.2 Cleaning: replace, fill, bin, and dummy variables


```python
df = df_raw.copy()

# "?" is not a number. Replace it with missing, then convert to numeric, before taking the mean.
df.replace("?", np.nan, inplace=True)
for column in ["horsepower", "stroke", "bore", "price", "engine-size", "highway-mpg"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")

df["horsepower"] = df["horsepower"].fillna(df["horsepower"].mean())
df["stroke"] = df["stroke"].fillna(df["stroke"].mean())
df["bore"] = df["bore"].fillna(df["bore"].mean())

# Fill a category column with its mode. In the doors exercise, missing values were filled with "four".
door_mode = df["num-of-doors"].mode().iloc[0]
df["num-of-doors"] = df["num-of-doors"].fillna(door_mode)

# Price is the target. Rows without it cannot be used for supervised learning, so drop them.
df.dropna(subset=["price"], axis=0, inplace=True)
df.reset_index(drop=True, inplace=True)

# Cut the continuous price into three bins so the distribution is easier to read.
bins = np.linspace(df["price"].min(), df["price"].max(), 4)
df["price-binned"] = pd.cut(df["price"], bins, labels=["Low", "Medium", "High"], include_lowest=True)

# Fuel type is text. A regression needs numeric columns, so expand it into dummy variables.
fuel_dummies = pd.get_dummies(df["fuel-type"], prefix="fuel")
df = pd.concat([df, fuel_dummies], axis=1)
df[["horsepower", "stroke", "num-of-doors", "price", "price-binned"]].head()
```

### 2.3 Exploration: correlation, regression plots, box plots, and groups


```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

numeric_cols = ["engine-size", "highway-mpg", "horsepower", "peak-rpm", "bore", "stroke", "price"]
print(df[numeric_cols].corr(method="pearson").round(2))

# A correlation near 1 or -1 means a stronger linear relationship. A small p-value means that correlation is unlikely to be a coincidence.
coef, p_value = stats.pearsonr(df["engine-size"], df["price"])
print(f"engine-size vs price: r={coef:.3f}, p={p_value:.3g}")

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
sns.regplot(x="engine-size", y="price", data=df, ax=axes[0])
sns.regplot(x="highway-mpg", y="price", data=df, ax=axes[1])
sns.boxplot(x="drive-wheels", y="price", data=df, ax=axes[2])
axes[0].set_title("engine-size")
axes[1].set_title("highway-mpg")
axes[2].set_title("drive-wheels")
plt.tight_layout()
plt.show()
```

```python
# Group means: first by drive wheels, then crossed with body style.
by_drive = df.groupby("drive-wheels", as_index=False)[["price", "horsepower"]].mean()
print(by_drive.round(1))

pivot = df.pivot_table(values="price", index="drive-wheels", columns="body-style", aggfunc="mean")
plt.figure(figsize=(6, 3.5))
sns.heatmap(pivot, annot=True, cmap="RdBu", fmt=".0f")
plt.title("Mean price by drive-wheels and body-style")
plt.tight_layout()
plt.show()

plt.figure(figsize=(5, 3))
plt.hist(df["horsepower"], bins=3)
plt.xlabel("horsepower")
plt.ylabel("count")
plt.title("horsepower bins")
plt.tight_layout()
plt.show()
```

### 2.4 Models: simple, multiple, and a polynomial pipeline


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

y = df["price"]

# Simple linear regression: explain price with highway-mpg.
lm = LinearRegression()
lm.fit(df[["highway-mpg"]], y)
yhat_simple = lm.predict(df[["highway-mpg"]])
print("simple intercept:", round(lm.intercept_, 2))
print("simple slope:", round(lm.coef_[0], 2))
print("simple R2:", round(r2_score(y, yhat_simple), 3))

# Multiple linear regression: use horsepower, engine size, and fuel economy together.
features = ["horsepower", "engine-size", "highway-mpg"]
Z = df[features]
lm_multi = LinearRegression().fit(Z, y)
yhat_multi = lm_multi.predict(Z)
print("multiple R2:", round(r2_score(y, yhat_multi), 3))
print("multiple MSE:", round(mean_squared_error(y, yhat_multi), 1))

# Residual plot: a random scatter supports a linear fit. A curve is a reason to try a polynomial.
plt.figure(figsize=(5, 3.5))
sns.residplot(x=df["highway-mpg"], y=df["price"])
plt.title("Residual plot: highway-mpg")
plt.tight_layout()
plt.show()
```

```python
# A pipeline chains scaling, polynomial expansion, and linear regression so a transform is not skipped.
pipe = Pipeline([
    ("scale", StandardScaler()),
    ("polynomial", PolynomialFeatures(degree=2, include_bias=False)),
    ("model", LinearRegression()),
])
pipe.fit(Z, y)
yhat_pipe = pipe.predict(Z)
print("polynomial pipeline R2:", round(r2_score(y, yhat_pipe), 3))

plt.figure(figsize=(5, 3.5))
sns.kdeplot(y, color="r", label="Actual")
sns.kdeplot(yhat_multi, color="b", label="Fitted")
plt.xlabel("price")
plt.title("Actual vs fitted price")
plt.legend()
plt.tight_layout()
plt.show()
```

### 2.5 Evaluation: split, cross-validation, polynomial order, and Ridge


```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV, cross_val_score, train_test_split

x_train, x_test, y_train, y_test = train_test_split(Z, y, test_size=0.2, random_state=1)

# Fit on the training set and judge generalization on the test set. Training R2 alone is optimistic.
lm_split = LinearRegression().fit(x_train, y_train)
print("train R2:", round(lm_split.score(x_train, y_train), 3))
print("test R2:", round(lm_split.score(x_test, y_test), 3))

# Four-fold cross-validation rotates the training rows through the validation role, so the score is more stable.
cv_scores = cross_val_score(LinearRegression(), x_train, y_train, cv=4)
print("CV R2:", np.round(cv_scores, 3), "mean:", round(cv_scores.mean(), 3))

# A higher order fits the training data more closely. Test R2 does not have to keep rising.
r2_test = []
for degree in range(1, 5):
    poly = PolynomialFeatures(degree=degree, include_bias=False)
    xtr = poly.fit_transform(x_train)
    xte = poly.transform(x_test)
    model = LinearRegression().fit(xtr, y_train)
    r2_test.append(r2_score(y_test, model.predict(xte)))

plt.figure(figsize=(5, 3))
plt.plot(range(1, 5), r2_test, marker="o")
plt.xlabel("polynomial order")
plt.ylabel("test R2")
plt.title("R2 using test data")
plt.tight_layout()
plt.show()
```

```python
# Ridge uses alpha to penalize large coefficients. A larger alpha pulls the coefficients toward 0.
ridge = Ridge()
grid = GridSearchCV(ridge, {"alpha": [0.01, 0.1, 1, 10, 100]}, cv=4)
grid.fit(x_train, y_train)
print("best alpha:", grid.best_params_)
print("best CV R2:", round(grid.best_score_, 3))
print("test R2:", round(grid.score(x_test, y_test), 3))
```

## 3. Machine Learning with Python

Supervised models predict a number or a class from labeled data. Unsupervised models have no labels; they find groups or compress features. The last subsection puts evaluation, regularization, and tuning together.

The examples use data built into scikit-learn. They cover the lab topics: linear and logistic regression, trees, KNN, SVM, random forests, clustering, PCA, t-SNE, and Pipeline / GridSearchCV.


### 3.1 Supervised learning: linear and logistic regression


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_regression
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import classification_report, r2_score
from sklearn.model_selection import train_test_split

# Simple linear regression. make_regression builds one feature and a noisy target.
X_simple, y_simple = make_regression(n_samples=80, n_features=1, noise=15, random_state=0)
Xtr, Xte, ytr, yte = train_test_split(X_simple, y_simple, test_size=0.2, random_state=0)
simple = LinearRegression().fit(Xtr, ytr)
print("simple linear test R2:", round(r2_score(yte, simple.predict(Xte)), 3))

# Multiple linear regression: fit several features at once.
X_multi, y_multi = make_regression(n_samples=120, n_features=4, noise=20, random_state=0)
Xtr, Xte, ytr, yte = train_test_split(X_multi, y_multi, test_size=0.2, random_state=0)
multi = LinearRegression().fit(Xtr, ytr)
print("multiple linear test R2:", round(multi.score(Xte, yte), 3))
print("coefficients:", np.round(multi.coef_, 2))
```

```python
# Logistic regression classifies. It outputs class probabilities, not a continuous price.
iris = load_iris()
Xtr, Xte, ytr, yte = train_test_split(
    iris.data, iris.target, test_size=0.25, random_state=0, stratify=iris.target
)
logit = LogisticRegression(max_iter=200)
logit.fit(Xtr, ytr)
print("logistic accuracy:", round(logit.score(Xte, yte), 3))
print(classification_report(yte, logit.predict(Xte), target_names=iris.target_names))
```

### 3.2 Trees, KNN, SVM, random forests, and XGBoost


```python
from sklearn.datasets import load_iris, make_regression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor

iris = load_iris()
Xtr, Xte, ytr, yte = train_test_split(
    iris.data, iris.target, test_size=0.25, random_state=0, stratify=iris.target
)

# A decision tree splits on feature thresholds. max_depth limits the depth and reduces overfitting.
tree_clf = DecisionTreeClassifier(max_depth=3, random_state=0)
tree_clf.fit(Xtr, ytr)
print("decision tree accuracy:", round(tree_clf.score(Xte, yte), 3))

# KNN lets the k nearest neighbors vote. Standardize features first when their scales differ.
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(Xtr, ytr)
print("knn accuracy:", round(knn.score(Xte, yte), 3))

# An SVM looks for the separating surface with the largest margin. The RBF kernel handles a nonlinear boundary.
svm = SVC(kernel="rbf", random_state=0)
svm.fit(Xtr, ytr)
print("svm accuracy:", round(svm.score(Xte, yte), 3))

# A regression tree predicts a continuous value. The labs used it for amounts such as tips and house prices.
X_reg, y_reg = make_regression(n_samples=150, n_features=3, noise=12, random_state=1)
Xtr_r, Xte_r, ytr_r, yte_r = train_test_split(X_reg, y_reg, test_size=0.2, random_state=1)
tree_reg = DecisionTreeRegressor(max_depth=4, random_state=1).fit(Xtr_r, ytr_r)
pred = tree_reg.predict(Xte_r)
print("regression tree R2:", round(r2_score(yte_r, pred), 3))
print("regression tree MSE:", round(mean_squared_error(yte_r, pred), 2))
```

```python
from sklearn.ensemble import RandomForestClassifier

# A random forest is a vote among many trees. n_estimators is the number of trees.
forest = RandomForestClassifier(n_estimators=100, random_state=0)
forest.fit(Xtr, ytr)
print("random forest accuracy:", round(forest.score(Xte, yte), 3))

# Feature importance: a larger value means the trees used that column more often to split.
importances = forest.feature_importances_
order = np.argsort(importances)
plt.figure(figsize=(6, 3))
plt.barh(np.array(iris.feature_names)[order], importances[order], color="skyblue")
plt.xlabel("importance")
plt.title("Random forest feature importance")
plt.tight_layout()
plt.show()

try:
    from xgboost import XGBClassifier
    xgb = XGBClassifier(
        n_estimators=80,
        max_depth=3,
        random_state=0,
        eval_metric="mlogloss",
    )
    xgb.fit(Xtr, ytr)
    print("xgboost accuracy:", round(xgb.score(Xte, yte), 3))
except ImportError:
    print("xgboost is not installed; this step is skipped")
```

### 3.3 Unsupervised learning: K-Means, DBSCAN, PCA, t-SNE, and UMAP


```python
from sklearn.cluster import DBSCAN, KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import davies_bouldin_score, silhouette_score

X_blob, y_blob = make_blobs(n_samples=300, centers=4, cluster_std=0.7, random_state=0)

# K-Means needs the number of clusters in advance. A silhouette score closer to 1 means clearer clusters.
kmeans = KMeans(n_clusters=4, n_init=10, random_state=0)
labels_km = kmeans.fit_predict(X_blob)
print("k-means silhouette:", round(silhouette_score(X_blob, labels_km), 3))
print("k-means Davies-Bouldin:", round(davies_bouldin_score(X_blob, labels_km), 3))

# DBSCAN forms clusters from density. It can label sparse points as noise (-1) without a preset cluster count.
dbscan = DBSCAN(eps=0.8, min_samples=5)
labels_db = dbscan.fit_predict(X_blob)
n_clusters = len(set(labels_db) - {-1})
print("dbscan clusters:", n_clusters, "noise points:", int(np.sum(labels_db == -1)))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].scatter(X_blob[:, 0], X_blob[:, 1], c=labels_km, cmap="tab10", s=18)
axes[0].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c="red", marker="X", s=80)
axes[0].set_title("K-Means")
axes[1].scatter(X_blob[:, 0], X_blob[:, 1], c=labels_db, cmap="tab10", s=18)
axes[1].set_title("DBSCAN")
plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

iris = load_iris()
X_scaled = StandardScaler().fit_transform(iris.data)

# PCA finds the directions with the most variance. The first two components are enough for a scatter plot and show how much variance they explain.
pca = PCA(n_components=2, random_state=0)
X_pca = pca.fit_transform(X_scaled)
print("PCA explained variance ratio:", np.round(pca.explained_variance_ratio_, 3))

# t-SNE emphasizes local neighborhoods. It is useful for visualization and is a poor reusable transform for predicting new samples.
X_tsne = TSNE(n_components=2, random_state=0, perplexity=30).fit_transform(X_scaled)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=iris.target, cmap="tab10", s=20)
axes[0].set_title("PCA")
axes[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=iris.target, cmap="tab10", s=20)
axes[1].set_title("t-SNE")
plt.tight_layout()
plt.show()

try:
    import umap
    X_umap = umap.UMAP(n_components=2, random_state=0).fit_transform(X_scaled)
    plt.figure(figsize=(4.5, 3.5))
    plt.scatter(X_umap[:, 0], X_umap[:, 1], c=iris.target, cmap="tab10", s=20)
    plt.title("UMAP")
    plt.tight_layout()
    plt.show()
except ImportError:
    print("umap-learn is not installed; UMAP step is skipped")
```

### 3.4 Evaluation, regularization, Pipeline, and GridSearchCV


```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
from sklearn.model_selection import train_test_split

# Do not judge a classifier by accuracy alone. A confusion matrix shows which classes are mistaken for each other.
X_cls, y_cls = make_classification(
    n_samples=240, n_features=6, n_informative=3, n_classes=2, random_state=0
)
Xtr, Xte, ytr, yte = train_test_split(X_cls, y_cls, test_size=0.25, random_state=0, stratify=y_cls)
logit = LogisticRegression(max_iter=200).fit(Xtr, ytr)
y_pred = logit.predict(Xte)
print(confusion_matrix(yte, y_pred))
ConfusionMatrixDisplay.from_predictions(yte, y_pred)
plt.title("Logistic regression confusion matrix")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import Lasso, LinearRegression, Ridge
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Regularization: Ridge shrinks coefficients. Lasso can also push unimportant coefficients to 0.
X_reg, y_reg = make_regression(n_samples=120, n_features=8, n_informative=3, noise=15, random_state=0)
Xtr, Xte, ytr, yte = train_test_split(X_reg, y_reg, test_size=0.3, random_state=0)
# Transform the test set with the training mean and standard deviation so information does not leak.
scaler = StandardScaler().fit(Xtr)
Xtr_s = scaler.transform(Xtr)
Xte_s = scaler.transform(Xte)

linear = LinearRegression().fit(Xtr_s, ytr)
ridge = Ridge(alpha=1).fit(Xtr_s, ytr)
lasso = Lasso(alpha=0.5).fit(Xtr_s, ytr)
print("linear test R2:", round(linear.score(Xte_s, yte), 3))
print("ridge test R2:", round(ridge.score(Xte_s, yte), 3))
print("lasso test R2:", round(lasso.score(Xte_s, yte), 3))
print("lasso nonzero coefficients:", int(np.sum(np.abs(lasso.coef_) > 1e-6)))
```

```python
from sklearn.datasets import make_classification
from sklearn.decomposition import PCA
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Pipeline: standardize, then PCA, then KNN. GridSearchCV chooses the PCA dimension and the neighbor count inside cross-validation.
X_cls, y_cls = make_classification(
    n_samples=240, n_features=6, n_informative=3, n_classes=2, random_state=0
)
Xtr, Xte, ytr, yte = train_test_split(
    X_cls, y_cls, test_size=0.25, random_state=0, stratify=y_cls
)
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA()),
    ("knn", KNeighborsClassifier()),
])
param_grid = {
    "pca__n_components": [2, 3],
    "knn__n_neighbors": [3, 5, 7],
}
search = GridSearchCV(
    pipeline,
    param_grid,
    cv=StratifiedKFold(n_splits=4, shuffle=True, random_state=0),
    scoring="accuracy",
)
search.fit(Xtr, ytr)
print("best params:", search.best_params_)
print("best CV accuracy:", round(search.best_score_, 3))
print("test accuracy:", round(search.score(Xte, yte), 3))
```

## 4. SQL

Use Python's built-in `sqlite3` to create tables, insert rows, query, and update. Then read the results into pandas.

The labs also used the `%%sql` magic command. That magic needs a separate `ipython-sql` install, so the same SQL is sent through `cursor.execute` and `pandas.read_sql_query`. The connection is an in-memory database, so closing the notebook does not leave a `.db` file.


### 4.1 Create, insert, query, and update


```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS INSTRUCTOR")
cursor.execute(
    """
    CREATE TABLE IF NOT EXISTS INSTRUCTOR (
        ID INTEGER PRIMARY KEY NOT NULL,
        FNAME VARCHAR(20),
        LNAME VARCHAR(20),
        CITY VARCHAR(20),
        CCODE CHAR(2)
    )
    """
)

cursor.execute("INSERT INTO INSTRUCTOR VALUES (1, 'Rav', 'Ahuja', 'TORONTO', 'CA')")
cursor.execute(
    """
    INSERT INTO INSTRUCTOR VALUES
        (2, 'Raul', 'Chong', 'Markham', 'CA'),
        (3, 'Hima', 'Vasudevan', 'Chicago', 'US')
    """
)
conn.commit()
print("Table is Ready")

cursor.execute("SELECT * FROM INSTRUCTOR")
print("fetchall:")
for row in cursor.fetchall():
    print(row)

cursor.execute("SELECT * FROM INSTRUCTOR")
print("fetchmany(2):")
for row in cursor.fetchmany(2):
    print(row)

cursor.execute("SELECT FNAME FROM INSTRUCTOR")
print("first names:", cursor.fetchall())

cursor.execute("UPDATE INSTRUCTOR SET CITY = 'MOOSETOWN' WHERE FNAME = 'Rav'")
conn.commit()
cursor.execute("SELECT FNAME, CITY FROM INSTRUCTOR WHERE FNAME = 'Rav'")
print("after update:", cursor.fetchone())
```

### 4.2 Read query results with pandas


```python
# read_sql_query returns a DataFrame, so the filters and column selection from earlier still apply.
instructors = pd.read_sql_query("SELECT * FROM INSTRUCTOR", conn)
print(instructors)
print("shape:", instructors.shape)
print("first last name:", instructors.loc[0, "LNAME"])

canada = pd.read_sql_query("SELECT * FROM INSTRUCTOR WHERE CCODE = 'CA'", conn)
print(canada)
```

### 4.3 Aggregation, subqueries, and joins


```python
# Two small tables practice the lab patterns: GROUP BY, LIKE, ORDER BY, LIMIT, subqueries, and joins.
cursor.execute("DROP TABLE IF EXISTS SCHOOLS")
cursor.execute("DROP TABLE IF EXISTS CRIME")
cursor.execute(
    """
    CREATE TABLE SCHOOLS (
        NAME TEXT,
        AREA TEXT,
        LEVEL TEXT,
        SAFETY_SCORE INTEGER,
        ATTENDANCE TEXT,
        ENROLLMENT INTEGER
    )
    """
)
cursor.execute(
    """
    CREATE TABLE CRIME (
        CASE_NUMBER INTEGER,
        AREA TEXT,
        PRIMARY_TYPE TEXT,
        LOCATION TEXT
    )
    """
)
cursor.executemany(
    "INSERT INTO SCHOOLS VALUES (?, ?, ?, ?, ?, ?)",
    [
        ("North Elementary", "Uptown", "ES", 90, "96%", 400),
        ("Lake Middle", "Uptown", "MS", 70, "88%", 500),
        ("South High", "Downtown", "HS", 55, "68%", 800),
        ("West Elementary", "Westside", "ES", 40, "72%", 220),
        ("East High", "Downtown", "HS", 81, "91%", 640),
    ],
)
cursor.executemany(
    "INSERT INTO CRIME VALUES (?, ?, ?, ?)",
    [
        (1, "Downtown", "THEFT", "street"),
        (2, "Downtown", "BATTERY", "school yard"),
        (3, "Uptown", "KIDNAPPING", "apartment"),
        (4, "Downtown", "THEFT", "school parking"),
        (5, "Westside", "ASSAULT", "school hallway"),
        (6, "Uptown", "THEFT", "store"),
    ],
)
conn.commit()


def show(title, sql):
    print("\n" + title)
    print(pd.read_sql_query(sql, conn))


show("tables", "SELECT name FROM sqlite_master WHERE type = 'table'")
show(
    "elementary schools",
    "SELECT COUNT(*) AS elementary_schools FROM SCHOOLS WHERE LEVEL = 'ES'",
)
show(
    "highest safety score",
    """
    SELECT NAME, SAFETY_SCORE
    FROM SCHOOLS
    WHERE SAFETY_SCORE = (SELECT MAX(SAFETY_SCORE) FROM SCHOOLS)
    """,
)
show(
    "attendance as a number",
    """
    SELECT NAME,
           CAST(REPLACE(ATTENDANCE, '%', '') AS DECIMAL) AS attendance_pct
    FROM SCHOOLS
    WHERE CAST(REPLACE(ATTENDANCE, '%', '') AS DECIMAL) < 70
    ORDER BY attendance_pct
    """,
)
show(
    "enrollment by area",
    """
    SELECT AREA, SUM(ENROLLMENT) AS total_enrollment
    FROM SCHOOLS
    GROUP BY AREA
    ORDER BY total_enrollment
    LIMIT 5
    """,
)
show(
    "kidnapping cases",
    "SELECT CASE_NUMBER, PRIMARY_TYPE FROM CRIME WHERE PRIMARY_TYPE LIKE '%KIDNAPPING%'",
)
show(
    "cases at school locations",
    """
    SELECT PRIMARY_TYPE, COUNT(*) AS number_of_cases
    FROM CRIME
    WHERE LOCATION LIKE '%school%'
    GROUP BY PRIMARY_TYPE
    """,
)
show(
    "area with the most cases",
    """
    SELECT S.AREA, COUNT(*) AS cases
    FROM SCHOOLS AS S
    JOIN CRIME AS C ON S.AREA = C.AREA
    GROUP BY S.AREA
    ORDER BY cases DESC
    LIMIT 1
    """,
)

conn.close()
```
